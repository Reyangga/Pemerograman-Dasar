```python
#pratikum 4
#pertemuan 7
#break saat menemukan angka -1
data_angka = [10, 20, 30, -1, 50, 60]

print("Memproses data:")
for angka in data_angka:
    if angka == -1:
        print("  → Angka -1 ditemukan! Proses dihentikan.")
        break
    print(f"  Memproses: {angka}")

print("Program selesai.")

#break saat menemukan angka 0
data_sensor = [45.2, 67.8, 89.1, 0, 55.3, 72.4]

print("Membaca data sensor pembangkit:")
for nilai in data_sensor:
    if nilai == 0:
        print("  ⚠ Pembacaan sensor = 0! Kemungkinan sensor mati. Proses dihentikan.")
        break
    print(f"  Suhu terbaca: {nilai}°C")
#continue untuk melewati angka negatif
data_daya = [120.5, -1, 135.2, 0.0, 98.7, -2, 145.0]

print("Data daya valid (MW):")
for daya in data_daya:
    if daya <= 0:
        print(f"  (Melewati data tidak valid: {daya})")
        continue
    print(f"  Daya: {daya} MW")

#penambahan berkelanjutan total >20
total = 0
bilangan = 1
batas = 20

while total <= batas:
    total += bilangan
    bilangan += 1

print(f"Total pertama yang melebihi {batas}: {total}")
print(f"Bilangan terakhir ditambahkan: {bilangan - 1}")
```

```python
#pratikum 5
#pertemuan 7
#mencetak pyton 3 baris

jumlah_baris = 3

for baris in range(jumlah_baris):
    baris_teks = ""
    for _ in range(1):
        baris_teks += "Python "
    print(baris_teks.strip())
#pola segitiga bintang
print("Pola Segitiga:")
for baris in range(1, 6):
    for kolom in range(baris):
        print("*", end=" ")
    print()

#tabel suhu sensor
nama_unit = ["Unit 1", "Unit 2", "Unit 3"]
suhu_harian = [
    [85.2, 87.1, 86.5, 88.0, 84.9, 85.7, 86.2],   # Unit 1
    [90.1, 91.3, 89.7, 92.0, 90.5, 91.1, 90.8],   # Unit 2
    [78.4, 79.2, 77.8, 80.1, 79.5, 78.9, 79.7],   # Unit 3
]
hari = ["Sen", "Sel", "Rab", "Kam", "Jum", "Sab", "Min"]

print("Laporan Suhu Boiler (°C)")
print(f"{'Unit':<10}", end="")
for h in hari:
    print(f"{h:>6}", end="")
print(f"{'Rata2':>8}")
print("-" * 60)

for i, unit in enumerate(nama_unit):   # outer loop: per unit
    print(f"{unit:<10}", end="")
    total = 0
    for suhu in suhu_harian[i]:         # inner loop: per hari
        print(f"{suhu:>6.1f}", end="")
        total += suhu
    rata2 = total / len(suhu_harian[i])
    print(f"{rata2:>8.1f}")
```

```python
#latihan pratikum
#soal 2
data_daya = [
    [142, 138, 145, 140],
    [115, 120, 118, 122],
    [95, 98, 92, 96],
]
nama_unit = ["Unit A", "Unit B", "Unit C"]

print("--- Soal L2: Rata-rata Daya per Unit ---")
for i in range(len(data_daya)):
    total = 0
    for daya in data_daya[i]:
        total += daya
    rata2 = total / len(data_daya[i])
    print(f"{nama_unit[i]}: {rata2:.1f} MW")
```

```python
#latihan pratikum
#soal 3
print("--- Soal L3: Monitoring Sensor ---")
import random
random.seed(42)

siklus = 0
ada_anomali = True

while ada_anomali:
    siklus += 1
    ada_anomali = False
    print(f"\nSiklus ke-{siklus}:")
    for sensor_id in range(1, 5):
        suhu = random.randint(65, 95)
        if suhu < 70 or suhu > 90:
            print(f"  ⚠ Sensor {sensor_id}: {suhu}°C — ANOMALI")
            ada_anomali = True
        else:
            print(f"  ✓ Sensor {sensor_id}: {suhu}°C — Normal")

print(f"\nSemua sensor normal pada siklus ke-{siklus}.")
```
