```python
scores = [85, 70, 95, 80, 60]

total = 0                    # accumulator: dimulai dari nol
count_pass = 0               # counter: menghitung jumlah nilai yang memenuhi syarat
for score in scores:
    total = total + score    # menjumlahkan setiap elemen
    if score >= 80:
        count_pass = count_pass + 1
    print("total sementara:", total)

average = total / len(scores)
print("Total          :", total)
print("Rata-rata      :", average)
print("Nilai >= 80    :", count_pass, "dari", len(scores))
```

```python
scores = [88, 72, 95, 68, 82]

highest = scores[0]          # nilai tertinggi sementara
lowest = scores[0]           # nilai terendah sementara
for score in scores:
    if score > highest:
        highest = score
    if score < lowest:
        lowest = score

print("Tertinggi:", highest)
print("Terendah :", lowest)
print("Pengecekan bawaan:", max(scores), min(scores))
```

```python
temps = [26, 32, 29, 35, 30]    # suhu (°C), hari ke-1 sampai hari ke-5

pos_max = 0                      # posisi nilai tertinggi sementara
for i in range(len(temps)):
    if temps[i] > temps[pos_max]:
        pos_max = i              # menyimpan POSISI, bukan hanya nilainya

print("Suhu tertinggi:", temps[pos_max], "°C")
print("Ditemukan pada indeks", pos_max, "= hari ke-", pos_max + 1)
```

```python
temps = [25, 30, 34, 29, 36]
target = 31                       # hari pertama dengan suhu lebih dari 31 °C?

found = None                      # None = belum ditemukan
checks = 0
for i in range(len(temps)):
    checks = checks + 1
    if temps[i] > target:
        found = i
        break                     # berhenti pada kecocokan pertama

print("Hari pertama di atas", target, "°C: indeks", found)
print("Jumlah elemen diperiksa:", checks, "dari", len(temps))

# Fungsi tambahan
print(34 in temps)                # apakah 34 ada di dalam list?
print(temps.index(34))            # posisi kemunculan pertama angka 34
```

```python
temps = [24, 29, 27, 34, 31, 26, 33]

# Perubahan suhu harian: mulai dari indeks 1 karena menggunakan i - 1
for i in range(1, len(temps)):
    change = temps[i] - temps[i - 1]
    print(f"hari {i + 1}: {temps[i]} °C, perubahan {change:+d}")

# Rata-rata bergerak 3 hari: mengambil 3 nilai hingga indeks i
avg3 = []
for i in range(2, len(temps)):
    window = temps[i - 2 : i + 1]
    avg3.append(round(sum(window) / 3, 1))

print("Rata-rata 3 hari:", avg3)
print(len(temps), "nilai ->", len(avg3), "rata-rata")
```

```python
energy = [
    5200, 5400, 5350, 5500, 5450, 5300, 5250, 5600, 5550, 5400,
    5500, 5350, 5450, 5200, 5300, 5600, 3800, 5500, 5550, 5400,
    5250, 5300, 5600, 5650, 5500, 5350, 5450, 5200, 5300, 5550,
]

total = 0
i_max = 0
i_min = 0
for i in range(len(energy)):
    total = total + energy[i]           # accumulator
    if energy[i] > energy[i_max]:
        i_max = i                        # menyimpan POSISI, bukan hanya nilai
    if energy[i] < energy[i_min]:
        i_min = i

print(f"Energi bulan ini : {total:,} MWh")
print(f"Rata-rata        : {total / len(energy):,.0f} MWh/hari")
print(f"Tertinggi        : hari ke-{i_max + 1} ({energy[i_max]:,} MWh)")
print(f"Terendah         : hari ke-{i_min + 1} ({energy[i_min]:,} MWh)")
print("Fungsi bawaan    :", sum(energy), max(energy), min(energy))
```

```python
import csv

# --- Reading the data (provided) -> parallel arrays ---
dates, energy, run_hours = [], [], []
with open("daily_production_synthetic.csv") as f:
    for row in csv.DictReader(f):
        dates.append(row["date"])
        energy.append(int(row["energy_mwh"]))
        run_hours.append(int(row["run_hours"]))

INSTALLED_MW = 330
DAY_NAMES = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]  # 1 November 2026 is a Sunday

# 1. Total, average, monthly capacity factor
total = sum(energy)
cf = total / (INSTALLED_MW * 24 * len(energy)) * 100
print(f"November energy : {total:,} MWh  |  average {total / len(energy):,.0f} MWh/day")
print(f"Capacity factor : {cf:.1f} %")

# 2. Weekdays vs weekends (day names repeat every 7 indices)
weekday, weekend = [], []
for i in range(len(energy)):
    if DAY_NAMES[i % 7] in ("Sat", "Sun"):
        weekend.append(energy[i])
    else:
        weekday.append(energy[i])
print(f"Weekday average {sum(weekday) / len(weekday):,.0f} vs weekend {sum(weekend) / len(weekend):,.0f} MWh")

# 3. Days without a full 24 hours of operation
for i in range(len(energy)):
    if run_hours[i] < 24:
        print(f"{dates[i]} ({DAY_NAMES[i % 7]}): only {run_hours[i]} h in operation, {energy[i]:,} MWh")

# 4. Text bar chart: one block = 100 MWh above 3,000 MWh (so the differences show)
print()
for i in range(len(energy)):
    print(f"{i + 1:>2} {DAY_NAMES[i % 7]} {'█' * ((energy[i] - 3000) // 100)} {energy[i]:,}")
```

```python
#latihan laboran


energy = [
    5200, 5400, 5350, 5500, 5450, 5300, 5250, 5600, 5550, 5400,
    5500, 5350, 5450, 5200, 5300, 5600, 3800, 5500, 5550, 5400,
    5250, 5300, 5600, 5650, 5500, 5350, 5450, 5200, 5300, 5550,
]

# 1. Menghitung total dan rata-rata energi
total = sum(energy)
average = total / len(energy)

print("Total energi    :", total, "MWh")
print("Rata-rata energi:", average, "MWh/hari")

# 2. Mencari tiga hari tertinggi dan terendah
data = list(enumerate(energy, start=1))

highest = sorted(data, key=lambda x: x[1], reverse=True)[:3]
lowest = sorted(data, key=lambda x: x[1])[:3]

print("\nTiga hari tertinggi:")
for day, value in highest:
    print("Hari", day, ":", value, "MWh")

print("\nTiga hari terendah:")
for day, value in lowest:
    print("Hari", day, ":", value, "MWh")

# 3. Menghitung jumlah hari di bawah target
target = int(input("\nMasukkan target energi (MWh): "))

below = 0
for value in energy:
    if value < target:
        below = below + 1

print("Jumlah hari di bawah target:", below)

# 4. Rata-rata pergerakan 7 hari
limit = average * 0.97

print("\nHari dengan rata-rata 7 hari di bawah 97%:")
for i in range(6, len(energy)):
    window = energy[i - 6 : i + 1]
    avg7 = sum(window) / 7

    if avg7 < limit:
        print("Hari", i + 1, ":", round(avg7, 2), "MWh")

# 5. Menghitung total energi per minggu
print("\nTotal energi per minggu:")
for i in range(0, len(energy), 7):
    week = energy[i : i + 7]
    week_number = i // 7 + 1

    print("Minggu", week_number, ":", sum(week), "MWh")
```
