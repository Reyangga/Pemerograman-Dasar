```python
scores = [80, 65, 90, 75, 70]

total = 0                    # accumulator: starts at zero
count_pass = 0               # a counter is an accumulator that adds 1
for score in scores:
    total = total + score    # add each element
    if score >= 75:
        count_pass = count_pass + 1
    print("running total:", total)

average = total / len(scores)
print("Total   :", total)
print("Average :", average)
print("Scores >= 75:", count_pass, "of", len(scores))
```

```python
scores = [80, 65, 90, 75, 70]

highest = scores[0]          # best so far (assume the first one)
lowest = scores[0]
for score in scores:
    if score > highest:
        highest = score
    if score < lowest:
        lowest = score

print("Highest:", highest)
print("Lowest :", lowest)
print("Built-in check:", max(scores), min(scores))
```

```python
temps = [28, 31, 30, 33, 29]    # temperature (°C), day 1 to day 5

pos_max = 0                      # position of the best so far
for i in range(len(temps)):
    if temps[i] > temps[pos_max]:
        pos_max = i              # keep the POSITION, not just the value

print("Hottest value:", temps[pos_max], "°C")
print("Found at index", pos_max, "= day", pos_max + 1)
```

```python
temps = [28, 31, 30, 33, 29]
target = 32                       # first day hotter than 32 °C?

found = None                      # None = "nothing found (yet)"
checks = 0
for i in range(len(temps)):
    checks = checks + 1
    if temps[i] > target:
        found = i
        break                     # stop at the first match

print("First day above", target, "°C: index", found)
print("Elements checked:", checks, "of", len(temps))

# Handy shortcuts
print(33 in temps)                # is 33 somewhere in the list?
print(temps.index(33))            # position of the first 33 (error if absent)
```

```python
temps = [28, 31, 30, 33, 29]

hot = []                          # the new list starts empty
for t in temps:
    if t >= 30:
        hot.append(t)
print("Hot values  :", hot)
print("How many    :", len(hot))

# To keep the day numbers instead of the values, loop by position
hot_days = []
for i in range(len(temps)):
    if temps[i] >= 30:
        hot_days.append(i + 1)
print("Hot days    :", hot_days)

# Optional: the same filter as a list comprehension (a compact loop)
hot_again = [t for t in temps if t >= 30]
print("Same result?", hot_again == hot)
```

```python
temps = [28, 31, 30, 33, 29, 27, 32]

# Daily change: start at index 1 because we look at i - 1
for i in range(1, len(temps)):
    change = temps[i] - temps[i - 1]
    print(f"day {i + 1}: {temps[i]} °C, change {change:+d}")

# 3-day moving average: window = the 3 values ending at index i
avg3 = []
for i in range(2, len(temps)):
    window = temps[i - 2 : i + 1]
    avg3.append(round(sum(window) / 3, 1))

print("3-day averages:", avg3)
print(len(temps), "values ->", len(avg3), "averages")
```

```python
nums = [1, 2, 2, 3]

# Wrong: removing from the list while looping over it
for x in nums:
    if x == 2:
        nums.remove(x)
print("wrong:", nums)           # one 2 survived!

# Right: build a new list with only what you want to keep
nums = [1, 2, 2, 3]
kept = []
for x in nums:
    if x != 2:
        kept.append(x)
print("right:", kept)
```

```python
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

total = 0
i_max = 0
i_min = 0
for i in range(len(energy)):
    total = total + energy[i]           # accumulator
    if energy[i] > energy[i_max]:
        i_max = i                        # keep the POSITION, not just the value
    if energy[i] < energy[i_min]:
        i_min = i

print(f"Energy this month: {total:,} MWh")
print(f"Average          : {total / len(energy):,.0f} MWh/day")
print(f"Highest          : day {i_max + 1} ({energy[i_max]:,} MWh)")
print(f"Lowest           : day {i_min + 1} ({energy[i_min]:,} MWh)")
print("Built-ins        :", sum(energy), max(energy), min(energy))
```

```python
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]
target = 5800

# Linear search: first day below target
first_day = None
for i in range(len(energy)):
    if energy[i] < target:
        first_day = i + 1
        break
print("First day below target:", first_day)

# Filtering with an ordinary loop
short_days = []
for i in range(len(energy)):
    if energy[i] < target:
        short_days.append(i + 1)
print("Days below target:", short_days)

# The same filter as a list comprehension
short_days_2 = [i + 1 for i in range(len(energy)) if energy[i] < target]
print("Same result?", short_days == short_days_2)
print(f"Target met on {len(energy) - len(short_days)} of {len(energy)} days")
```

```python
import csv

# --- Reading the data (provided) -> parallel arrays ---
dates, energy, run_hours = [], [], []
with open("daily_production_synthetic.csv") as f:
    for row in csv.DictReader(f):
        dates.append(row["date"])
        energy.append(int(row["energy_mwh"]))
        run_hours.append(int(row["run_hours"]))

INSTALLED_MW = 330
DAY_NAMES = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]  # 1 November 2026 is a Sunday

# 1. Total, average, monthly capacity factor
total = sum(energy)
cf = total / (INSTALLED_MW * 24 * len(energy)) * 100
print(f"November energy : {total:,} MWh  |  average {total / len(energy):,.0f} MWh/day")
print(f"Capacity factor : {cf:.1f} %")

# 2. Weekdays vs weekends (day names repeat every 7 indices)
weekday, weekend = [], []
for i in range(len(energy)):
    if DAY_NAMES[i % 7] in ("Sat", "Sun"):
        weekend.append(energy[i])
    else:
        weekday.append(energy[i])
print(f"Weekday average {sum(weekday) / len(weekday):,.0f} vs weekend {sum(weekend) / len(weekend):,.0f} MWh")

# 3. Days without a full 24 hours of operation
for i in range(len(energy)):
    if run_hours[i] < 24:
        print(f"{dates[i]} ({DAY_NAMES[i % 7]}): only {run_hours[i]} h in operation, {energy[i]:,} MWh")

# 4. Text bar chart: one block = 100 MWh above 3,000 MWh (so the differences show)
print()
for i in range(len(energy)):
    print(f"{i + 1:>2} {DAY_NAMES[i % 7]} {'█' * ((energy[i] - 3000) // 100)} {energy[i]:,}")
```
