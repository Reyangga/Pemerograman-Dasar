```python
#Reyangga Alvonzo
#4232501021
#24/09/2026
#Membuat program untuk menghitung operasi aritmatika dua bilangan bulat
#Bsoal1 no1

print ("===OPERASI ARITMATIKA===")

bilangan1 = int(input("masukkan bilangan pertama:"))
bilangan2 = int(input("masukkan bilangan kedua:"))

print("/Hasil operasi:")
print(f"penjumlahan :{bilangan1 + bilangan2}")
print(f"pengurangan :{bilangan1 - bilangan2}")
print(f"perkalian :{bilangan1 * bilangan2}")
print(f"pembagian :{bilangan1 / bilangan2}")
print(f"pembagian bulat :{bilangan1 // bilangan2}")
print(f"modulus :{bilangan1 % bilangan2}")

#Reyangga Alvonzo
#4232501021
#24/09/2026
#Tujuan:Menerapkan operator relasional untuk menentukan apakah nilai ujian
#termasukkategori lulus
#soal B1 NO 2

print ("===OPERATOR RELASIONAL===")

nilai = int(input("masukkan nilai ujian (o-100):"))
hasil = nilai >=60
print (f"\nApakah nilai termasuk kategori lulus (>=60)?{hasil}")

#Reyangga Alvonzo
#4232501021
#24/09/2026
#Tujuan:Menerapka operator logika AND dan operator
#precedence untuk menentukan izin mengemudi
#soal B1 NO 3

print ("===OPERATOR LOGIKA===")

umur = int(input("masukkan umur:"))
sim = int(input("apakah memiliki SIM? (1 = Ya, 0 = Tidak):"))
punya_sim = sim == 1
boleh_mengemudi = (umur >=17) and punya_sim
print(f"Apakah boleh mengemudi sendiri?{boleh_mengemudi}")

#Reyangga Alvonzo
#4232501021
#24/09/2026
#Tujuan : Menghitung efisiensi termal
#soal B2 NO 4

print("===EFISIENSI TERMAL PEMBANGKIT===")

energi_input = float(input("Masukkan energi input(MJ):"))
energi_output = float(input("Masukkan energi output(MJ):"))
efisiensi = (energi_output / energi_input)*100
print(f"efisiensi termal pembangkit adalah{efisiensi:.2f}%")

#Reyangga Alvonzo
#4232501021
#24/09/2026
#Tujuan:Menentukan apakah unit pembangkit perlu maintenance

print("===KONDISI MAINTENANCE PEMBANGKIT===")

suhu_boiler = float(input("Masukkan suhu boiler (Celsius):"))
beban_operasi = float(input("Masukkan beban operasi (%):"))
perlu_maintenance = (suhu_boiler > 550) or(beban_operasi > 95)
print(f"Apakah unit pembangkit perlu maintenance?{perlu_maintenance}")
```
