```python
#no 3
suhu = [92.5, 88.7, 95.1, 102.3, 97.8]
print("Suhu titik ke-3 =", suhu[2], "°C")
print("Tiga pembacaan pertama =", suhu[:3])
suhu.append(99.4)
print("Jumlah data setelah penambahan =", len(suhu))
```

Kemudian, perintah print("Tiga pembacaan pertama =", suhu[:3]) digunakan untuk menampilkan tiga data suhu pertama dari list. Tanda [:3] berarti mengambil data mulai dari indeks 0 sampai sebelum indeks 3, sehingga hasilnya adalah 92.5, 88.7, dan 95.1.

Setelah itu, perintah suhu.append(99.4) digunakan untuk menambahkan data suhu baru sebesar 99.4 ke bagian akhir list. Fungsi append() berfungsi untuk menambahkan elemen baru ke dalam list tanpa menghapus data sebelumnya.

Pada proses terakhir, digunakan perintah len(suhu) untuk menghitung jumlah seluruh data yang ada dalam list setelah penambahan data baru. Karena sebelumnya ada lima data dan ditambahkan satu data lagi, maka jumlah keseluruhannya menjadi enam data.


```python
#no 5
tekanan_uap = float(input("Masukkan tekanan uap (psi): "))

status_manual_override = False

if tekanan_uap > 150 and status_manual_override == False:
    status = "Katup Terbuka"
elif status_manual_override == True:
    status = "Katup Mode Manual"
else:
    status = "Katup Tertutup"

print("Status Katup:", status)
```

Program ini digunakan untuk menentukan apakah sebuah katup dalam kondisi terbuka, tertutup, atau menggunakan mode manual berdasarkan tekanan uap. Pertama, program meminta pengguna memasukkan nilai tekanan uap dalam satuan PSI. Fungsi float() digunakan agar angka yang dimasukkan bisa dibaca sebagai angka desimal.

Selanjutnya, dibuat variabel status_manual_override dengan nilai False, yang artinya pengaturan manual sedang tidak aktif. Setelah itu, program mulai memeriksa kondisi menggunakan if. Jika tekanan uap lebih dari 150 PSI dan mode manual tidak aktif, maka status katup akan menjadi "Katup Terbuka".

Jika kondisi pertama tidak terpenuhi, program akan memeriksa kondisi berikutnya menggunakan elif. Jika mode manual bernilai True, maka katup akan masuk ke "Katup Mode Manual". Namun, jika kedua kondisi tersebut tidak terpenuhi, program akan menjalankan perintah else dan status katup menjadi "Katup Tertutup".

Terakhir, perintah print() digunakan untuk menampilkan hasil pemeriksaan ke layar. Jadi, program ini membantu menentukan kondisi katup secara otomatis berdasarkan tekanan uap yang dimasukkan dan apakah mode manual sedang aktif atau tidak.

```python
#no 7
suhu_saat_ini = 60

while suhu_saat_ini < 85:
    suhu_saat_ini += 5

    print(f"Suhu saat ini: {suhu_saat_ini} C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")
```

Program no 7 ini digunakan untuk menaikkan suhu secara bertahap sampai mencapai target 85°C. Pada awal program, dibuat variabel suhu_saat_ini = 60, yang berarti suhu awal berada di angka 60°C.

Selanjutnya, digunakan perintah while suhu_saat_ini < 85 untuk menjalankan perulangan selama suhu masih di bawah 85°C. Di dalam perulangan tersebut, terdapat perintah suhu_saat_ini += 5 yang berfungsi untuk menambahkan suhu sebanyak 5°C setiap kali perulangan berjalan.

Setelah suhu bertambah, perintah print() digunakan untuk menampilkan suhu terbaru ke layar. Kemudian, terdapat percabangan if suhu_saat_ini == 80 untuk memeriksa apakah suhu sudah mencapai 80°C. Jika kondisi tersebut terpenuhi, program akan menampilkan pesan "WARNING: Mendekati target!" sebagai peringatan bahwa suhu sudah mendekati target.

Perulangan akan terus berjalan sampai suhu mencapai 85°C. Ketika suhu sudah mencapai angka tersebut, kondisi while tidak lagi terpenuhi sehingga perulangan berhenti. Setelah itu, perintah print("Target 85 C tercapai.") dijalankan untuk menampilkan pesan bahwa target suhu sudah berhasil dicapai.

```python
#no 9
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

jumlah_tidak_valid = 0

for i in range(len(data)):
    nomor_unit = i + 1

    try:
        nilai = float(data[i])

    except ValueError:
        print(f"Unit {nomor_unit}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

    if nilai > 2.0:
        print(f"Warning: Unit {nomor_unit} vibrasi tinggi!")

    if nilai > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break

print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

Program ini digunakan untuk memeriksa data getaran dari beberapa unit mesin dan mengetahui apakah ada data yang tidak valid atau tingkat getarannya terlalu tinggi. Pada awal program, dibuat variabel data yang berisi lima data getaran dalam bentuk teks, yaitu 1.2, 0.9, ERR, 3.1, dan 1.8. Kemudian, dibuat variabel jumlah_tidak_valid = 0 untuk menghitung berapa banyak data yang tidak bisa dibaca atau tidak valid.

Selanjutnya, digunakan perulangan for i in range(len(data)) untuk memeriksa setiap data yang ada di dalam list. Fungsi len(data) digunakan untuk mengetahui jumlah data, sedangkan range() digunakan untuk membuat urutan indeks dari data tersebut. Setelah itu, dibuat variabel nomor_unit = i + 1 agar nomor unit dimulai dari angka 1, bukan 0.

Kemudian, digunakan try untuk mencoba mengubah data menjadi angka desimal menggunakan fungsi float(data[i]). Jika data berhasil diubah, program akan melanjutkan pemeriksaan. Namun, jika data tidak bisa diubah menjadi angka, seperti tulisan "ERR", maka bagian except ValueError akan dijalankan. Program akan menampilkan pesan bahwa data unit tersebut tidak valid, kemudian jumlah_tidak_valid += 1 digunakan untuk menambahkan jumlah data yang tidak valid sebanyak satu. Setelah itu, perintah continue membuat program melewati pemeriksaan data tersebut dan langsung lanjut ke unit berikutnya.

Jika data berhasil dibaca, program akan memeriksa apakah nilai getaran lebih dari 2.0. Jika iya, maka akan muncul pesan peringatan bahwa getaran unit tersebut tinggi. Selanjutnya, terdapat pemeriksaan kedua menggunakan if nilai > 3.0. Jika nilai getaran lebih dari 3.0, program akan menampilkan pesan kritis dan menjalankan perintah break untuk menghentikan seluruh perulangan karena kondisi getaran sudah dianggap berbahaya.

Pada bagian terakhir, perintah print() digunakan untuk menampilkan jumlah data yang tidak valid selama pemeriksaan berlangsung.


```python
#Soal tantangan
# Program Penghitungan Konsumsi Energi Listrik Harian
# Program menghitung energi listrik dari beberapa peralatan dan menentukan kategori konsumsi listrik berdasarkan total kWh.

nama_alat = []
daya_alat = []
lama_pemakaian = []

total_energi = 0

print("=== PROGRAM KONSUMSI ENERGI LISTRIK ===")

try:
    jumlah = int(input("Masukkan jumlah peralatan listrik: "))

    if jumlah <= 0:
        print("Jumlah peralatan harus lebih dari 0")

    else:

        for i in range(jumlah):
            print("\nPeralatan ke-", i + 1)

            nama = input("Nama peralatan: ")
            daya = float(input("Daya listrik (Watt): "))
            jam = float(input("Lama pemakaian (jam): "))

            if daya <= 0 or jam < 0 or jam > 24:
                print("Input daya atau waktu tidak valid!")
                break

            nama_alat.append(nama)
            daya_alat.append(daya)
            lama_pemakaian.append(jam)

        if len(nama_alat) == jumlah:

            print("\n=== HASIL PERHITUNGAN ===")

            for i in range(jumlah):
                energi = (daya_alat[i] * lama_pemakaian[i]) / 1000

                total_energi += energi

                print(nama_alat[i], "=", energi, "kWh")

            print("\nTotal konsumsi listrik:", total_energi, "kWh")


            if total_energi < 5:
                print("Kategori: Hemat")

            elif total_energi <= 10:
                print("Kategori: Sedang")

            else:
                print("Kategori: Tinggi")

except ValueError:
    print("Error: Masukkan angka yang valid!")
```

Program ini dibuat untuk menghitung konsumsi energi listrik harian dari beberapa peralatan rumah tangga, seperti TV LED dengan daya 100 Watt, kipas angin 50 Watt, dan AC 1/2 PK dengan daya 400 Watt. Kita diminta memasukkan jumlah peralatan, nama alat, daya listrik, serta lama waktu pemakaian dalam jam. Data yang dimasukkan akan disimpan menggunakan list, kemudian program menggunakan perulangan for untuk menghitung konsumsi energi setiap peralatan dengan rumus daya dikalikan waktu pemakaian lalu dibagi 1000 sehingga menghasilkan satuan kWh. Selanjutnya, hasil perhitungan setiap peralatan dijumlahkan untuk mendapatkan total konsumsi energi listrik. Program juga menggunakan percabangan if-elif-else untuk menentukan kategori penggunaan listrik, yaitu hemat, sedang, atau tinggi. Selain itu, try-except digunakan untuk menangani kesalahan apabila pengguna memasukkan data yang tidak sesuai. Dengan adanya program ini, pengguna dapat mengetahui jumlah konsumsi energi listrik harian dan lebih mudah mengontrol penggunaan listrik agar lebih efisien.
